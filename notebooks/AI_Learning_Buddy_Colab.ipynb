## 🌐 Step 5 — Expose the App using ngrok

Now we initialize `pyngrok` using our auth token and open a secure tunnel to port 8501. This returns a public URL we can click to access the app from any browser!

```python
import os
from dotenv import load_dotenv
from pyngrok import ngrok

# Load keys from .env
load_dotenv()
ngrok_token = os.getenv("NGROK_AUTH_TOKEN")

if ngrok_token and ngrok_token != "YOUR_NGROK_AUTH_TOKEN_HERE":
    ngrok.set_auth_token(ngrok_token)
    public_url = ngrok.connect(8501)
    print("\n🎉 CONGRATULATIONS! Your AI Learning Buddy is live at:")
    print(public_url.public_url)
    print("\nClick the link above to open your web application in a new tab.")
else:
    print("❌ ERROR: Please enter a valid NGROK_AUTH_TOKEN in the Step 3 cell first!")
```
